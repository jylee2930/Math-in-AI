```python
# 1. 문장 준비
sentences = [
    "Fast delivery and good quality",
    "I'm so disappointed",
    "This is a really great product I'm completely satisfied",
    "It's really the worst I'll never buy it again"
]

# 2. 단어 사전 만들기
vocabulary = [
    "delivery", "fast", "quality", "good", "so",
    "bad", "disappointed", "really", "great", "product",
    "completely", "satisfied", "worst", "buy", "never"
]

print("단어 사전:")
print(vocabulary)


# 3. One-Hot Encoding 함수 만들기
def one_hot_encode(sentence, vocabulary):
    sentence = sentence.lower()
    words = sentence.split()

    vector = []

    for word in vocabulary:
        if word in words:##단어사전에 있으면 1
            vector.append(1)
        else:##단어사전에 없으면 0
            vector.append(0)

    return vector


# 4. 각 문장을 One-Hot Encoding
vectors = []

for sentence in sentences:
    vector = one_hot_encode(sentence, vocabulary)
    vectors.append(vector)

print("\nOne-Hot Encoding 결과:")
for i, vector in enumerate(vectors):
    print(f"문장 {i+1}: {vector}")


# 5. 코사인 유사도 함수 만들기
import math

def cosine_similarity(vector1, vector2):
    dot_product = 0
    ##내적구하기
    for i in range(len(vector1)):
        dot_product += vector1[i] * vector2[i]
    ##베터 크기 게산
    size1 = math.sqrt(sum(x * x for x in vector1))
    size2 = math.sqrt(sum(x * x for x in vector2))

    if size1 == 0 or size2 == 0:
        return 0

    return dot_product / (size1 * size2)


# 6. 문장 1과 문장 3의 유사도 계산
similarity_1_3 = cosine_similarity(vectors[0], vectors[2])
print("\n문장 1과 문장 3의 코사인 유사도:")
print(similarity_1_3)


# 7. 모든 문장끼리 유사도 비교
print("\n전체 문장 유사도:")

for i in range(len(vectors)):
    for j in range(i + 1, len(vectors)):
        similarity = cosine_similarity(vectors[i], vectors[j])
        print(f"문장 {i+1} vs 문장 {j+1}: {similarity:.3f}")
```

# 요즘 AI Word2Vec사용해서 단어의 위치 표현
##1. 문장을 벡터로 변환
##2. PCA로 2차원 축소
##3. 문자 위치를 그래프로 변환

```python
# 필요한 라이브러리 설치
# pip install scikit-learn matplotlib

# 1. 라이브러리 불러오기
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt


# 2. 문장 준비
sentences = [
    "Fast delivery and good quality",
    "I'm so disappointed",
    "This is a really great product I'm completely satisfied",
    "It's really the worst I'll never buy it again"
]


# 3. BoW 벡터 생성
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(sentences)


# 4. 벡터 확인
print("단어 사전:")
print(vectorizer.get_feature_names_out())

print("\nBoW 벡터:")
print(X.toarray())


# 5. PCA로 2차원 축소
pca = PCA(n_components=2)

X_2d = pca.fit_transform(X.toarray())


# 6. 좌표 출력
print("\n2차원 좌표:")
for i, point in enumerate(X_2d):
    print(f"문장 {i+1}: x={point[0]:.2f}, y={point[1]:.2f}")


# 7. 그래프 그리기
plt.figure(figsize=(8,6))

for i, point in enumerate(X_2d):

    plt.scatter(point[0], point[1], s=100)

    plt.text(
        point[0] + 0.02,
        point[1] + 0.02,
        f"Sentence {i+1}",
        fontsize=10
    )

# 제목
plt.title("Sentence Position Visualization")

# x축, y축 이름
plt.xlabel("X")
plt.ylabel("Y")

# 격자
plt.grid(True)

# 그래프 출력
plt.show()
```
